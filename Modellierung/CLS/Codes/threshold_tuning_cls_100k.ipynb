# Threshold Tuning for `order = 1`

Self-contained notebook for reproducible threshold tuning on the Orange TEST prediction export only.

Constraints enforced in this notebook:
- reads only `cls_test_100k_predictions_gb_logreg.csv`
- uses `skiprows=[1, 2]` to remove the two Orange metadata rows
- uses only `order`, `Gradient Boosting (sklearn default) (1)`, and `Logistic Regression (1)` for tuning
- does not import any project modules
- does not touch validation data or retrain models

## 1. Set Up the Development Environment

The notebook uses only `pandas`, `numpy`, and `sklearn.metrics`, plus standard-library helpers for file paths and structured metadata.

```python
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    f1_score,
    log_loss,
    matthews_corrcoef,
    precision_score,
    recall_score,
    roc_auc_score,
)

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 140)
pd.set_option("display.max_rows", 20)

print("Notebook environment ready.")
```

## 2. Define Configuration and Constants

This block fixes the input file, output files, target, probability columns, and the exact threshold grid required for the analysis.

```python
BASE_DIR = Path().resolve() / "Modellierung"  # portable: VS Code Jupyter CWD = workspace root
INPUT_PATH = BASE_DIR / "cls_test_100k_predictions_gb_logreg.csv"
RESULTS_PATH = BASE_DIR / "threshold_tuning_cls_100k_results.csv"
AUC_LOGLOSS_PATH = BASE_DIR / "threshold_tuning_cls_100k_auc_logloss.csv"
BEST_MODELS_PATH = BASE_DIR / "threshold_tuning_cls_100k_best_models.csv"

TARGET_COLUMN = "order"
SKIPROWS = [1, 2]
EXPECTED_SHAPE = (100000, 36)
THRESHOLDS = np.round(np.arange(5, 81) / 100, 2)

assert len(THRESHOLDS) == 76, f"Expected 76 thresholds, got {len(THRESHOLDS)}"

print(f"Input path: {INPUT_PATH}")
print(f"Threshold count: {len(THRESHOLDS)}")
print(f"Threshold range: {THRESHOLDS[0]:.2f} to {THRESHOLDS[-1]:.2f}")
```

## 3. Implement Core Data Structures

The notebook keeps the model mapping explicit so only the correct positive-class probability columns can be used for threshold tuning.

```python
MODEL_METADATA = {
    "Gradient Boosting": {
        "prob_col": "Gradient Boosting (sklearn default) (1)",
        "hard_pred_col": "Gradient Boosting (sklearn default)",
        "negative_prob_col": "Gradient Boosting (sklearn default) (0)",
    },
    "Logistic Regression": {
        "prob_col": "Logistic Regression (1)",
        "hard_pred_col": "Logistic Regression",
        "negative_prob_col": "Logistic Regression (0)",
    },
}

RESULT_COLUMNS = [
    "model",
    "threshold",
    "auc",
    "logloss",
    "accuracy",
    "precision",
    "recall",
    "f1",
    "mcc",
    "tn",
    "fp",
    "fn",
    "tp",
]

print(f"Chosen target column: {TARGET_COLUMN}")
print("Chosen probability columns:")
for model_name, info in MODEL_METADATA.items():
    print(f"- {model_name}: {info['prob_col']}")
```

## 4. Build the Main Processing Function

These helpers load the Orange CSV, validate the selected columns, compute probability metrics, scan thresholds, and choose the recommended winner.

```python
def load_predictions(file_path: Path) -> pd.DataFrame:
    if not file_path.exists():
        raise FileNotFoundError(f"Input file not found: {file_path}")

    df_loaded = pd.read_csv(file_path, skiprows=SKIPROWS)
    print(f"Using load strategy: pd.read_csv(..., skiprows={SKIPROWS})")
    return df_loaded


def coerce_binary_target(series: pd.Series, column_name: str) -> pd.Series:
    target = pd.to_numeric(series, errors="raise").astype(int)
    unique_values = set(target.unique().tolist())
    if unique_values != {0, 1}:
        raise ValueError(f"{column_name} must contain only 0/1, found {sorted(unique_values)}")
    return target


def coerce_probability_columns(df_loaded: pd.DataFrame, model_metadata: dict) -> pd.DataFrame:
    probability_data = {}

    for model_name, info in model_metadata.items():
        required_columns = [info["prob_col"], info["negative_prob_col"], info["hard_pred_col"]]
        missing = [column for column in required_columns if column not in df_loaded.columns]
        if missing:
            raise KeyError(f"Missing expected columns for {model_name}: {missing}")

        positive_prob = pd.to_numeric(df_loaded[info["prob_col"]], errors="raise")
        negative_prob = pd.to_numeric(df_loaded[info["negative_prob_col"]], errors="raise")

        if positive_prob.isna().any() or negative_prob.isna().any():
            raise ValueError(f"Probability columns for {model_name} contain missing values")
        if not positive_prob.between(0, 1).all():
            raise ValueError(f"Positive probabilities for {model_name} are outside [0, 1]")
        if not negative_prob.between(0, 1).all():
            raise ValueError(f"Negative probabilities for {model_name} are outside [0, 1]")

        pair_sum = positive_prob + negative_prob
        if not np.allclose(pair_sum.to_numpy(), 1.0, atol=1e-6):
            raise ValueError(f"Probability pair for {model_name} does not sum to 1 within tolerance")

        probability_data[info["prob_col"]] = positive_prob

    return pd.DataFrame(probability_data)


def safe_log_loss(y_true: pd.Series, y_prob: pd.Series | np.ndarray) -> float:
    clipped = np.clip(np.asarray(y_prob, dtype=float), 1e-15, 1 - 1e-15)
    return float(log_loss(y_true, clipped, labels=[0, 1]))


def evaluate_thresholds(
    y_true: pd.Series,
    y_prob: pd.Series,
    model_name: str,
    auc_value: float,
    logloss_value: float,
    thresholds: np.ndarray,
) -> pd.DataFrame:
    rows = []

    for threshold in thresholds:
        y_pred = (y_prob >= threshold).astype(int)
        tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()

        rows.append(
            {
                "model": model_name,
                "threshold": float(threshold),
                "auc": auc_value,
                "logloss": logloss_value,
                "accuracy": accuracy_score(y_true, y_pred),
                "precision": precision_score(y_true, y_pred, zero_division=0),
                "recall": recall_score(y_true, y_pred, zero_division=0),
                "f1": f1_score(y_true, y_pred, zero_division=0),
                "mcc": matthews_corrcoef(y_true, y_pred),
                "tn": int(tn),
                "fp": int(fp),
                "fn": int(fn),
                "tp": int(tp),
            }
        )

    return pd.DataFrame(rows, columns=RESULT_COLUMNS)


def rank_by_f1(results: pd.DataFrame) -> pd.DataFrame:
    ranked = results.assign(threshold_distance=(results["threshold"] - 0.50).abs())
    ranked = ranked.sort_values(
        by=["f1", "mcc", "precision", "recall", "threshold_distance", "model"],
        ascending=[False, False, False, False, True, True],
    )
    return ranked.drop(columns=["threshold_distance"]).reset_index(drop=True)


def rank_by_mcc(results: pd.DataFrame) -> pd.DataFrame:
    ranked = results.assign(threshold_distance=(results["threshold"] - 0.50).abs())
    ranked = ranked.sort_values(
        by=["mcc", "f1", "precision", "recall", "threshold_distance", "model"],
        ascending=[False, False, False, False, True, True],
    )
    return ranked.drop(columns=["threshold_distance"]).reset_index(drop=True)


def best_row_per_model(results: pd.DataFrame) -> pd.DataFrame:
    best_rows = []
    for model_name, group in results.groupby("model", sort=False):
        best_rows.append(rank_by_f1(group).iloc[[0]])
    return pd.concat(best_rows, ignore_index=True)


def evaluate_precision_collapse(winner_row: pd.Series, default_row: pd.Series) -> dict:
    default_precision = float(default_row["precision"])
    winner_precision = float(winner_row["precision"])
    winner_recall = float(winner_row["recall"])
    default_recall = float(default_row["recall"])

    collapse_flag = False
    message = "No precision collapse detected relative to threshold 0.50."
    precision_ratio = np.nan

    if default_precision > 0:
        precision_ratio = winner_precision / default_precision
        if precision_ratio < 0.5 and winner_recall > default_recall:
            collapse_flag = True
            message = (
                "Precision dropped to less than half of the 0.50 baseline while recall increased. "
                "Review this candidate before operational use."
            )

    return {
        "precision_collapse_vs_050": collapse_flag,
        "precision_ratio_vs_050": precision_ratio,
        "precision_check_message": message,
    }
```

## 5. Add Input and Output Handling

This cell loads the TEST prediction export, prints the required inspection output, cleans the target and probability columns, and prepares the inputs for threshold tuning.

```python
df = load_predictions(INPUT_PATH)

print("df.shape:", df.shape)
print("df.columns.tolist():")
print(df.columns.tolist())

if df.shape != EXPECTED_SHAPE:
    raise ValueError(f"Loaded shape {df.shape} does not match expected shape {EXPECTED_SHAPE}")

if TARGET_COLUMN not in df.columns:
    raise KeyError(f"Target column not found: {TARGET_COLUMN}")

y_true = coerce_binary_target(df[TARGET_COLUMN], TARGET_COLUMN)
probability_df = coerce_probability_columns(df, MODEL_METADATA)

selected_columns = [TARGET_COLUMN] + [info["prob_col"] for info in MODEL_METADATA.values()]
analysis_df = pd.concat([y_true.rename(TARGET_COLUMN), probability_df], axis=1)

print(f"Chosen target column: {TARGET_COLUMN}")
print("Chosen probability columns:")
for column_name in selected_columns[1:]:
    print(f"- {column_name}")

print("\nSelected data preview:")
print(analysis_df.head().to_string(index=False))
```

## 6. Run a Minimal End-to-End Example

The full TEST prediction file is used as the reproducible end-to-end example for the threshold scan and output generation.

```python
run_started_at = pd.Timestamp.now("UTC")

auc_logloss_rows = []
result_frames = []

for model_name, info in MODEL_METADATA.items():
    y_prob = analysis_df[info["prob_col"]]
    auc_value = float(roc_auc_score(y_true, y_prob))
    logloss_value = safe_log_loss(y_true, y_prob)

    auc_logloss_rows.append(
        {
            "model": model_name,
            "probability_column": info["prob_col"],
            "auc": auc_value,
            "logloss": logloss_value,
        }
    )
    result_frames.append(evaluate_thresholds(y_true, y_prob, model_name, auc_value, logloss_value, THRESHOLDS))

auc_logloss_df = pd.DataFrame(auc_logloss_rows).sort_values(by="auc", ascending=False).reset_index(drop=True)
results_df = pd.concat(result_frames, ignore_index=True)
results_df["threshold"] = results_df["threshold"].round(2)

if results_df.shape[0] != len(MODEL_METADATA) * len(THRESHOLDS):
    raise ValueError(
        f"Expected {len(MODEL_METADATA) * len(THRESHOLDS)} threshold rows, got {results_df.shape[0]}"
    )

top10_f1_df = rank_by_f1(results_df).head(10).reset_index(drop=True)
top10_mcc_df = rank_by_mcc(results_df).head(10).reset_index(drop=True)
default_threshold_df = results_df[np.isclose(results_df["threshold"], 0.50)].copy()
default_threshold_df = rank_by_f1(default_threshold_df)

best_models_df = best_row_per_model(results_df)
overall_winner = rank_by_f1(results_df).iloc[0].copy()
overall_winner_default = default_threshold_df.loc[default_threshold_df["model"] == overall_winner["model"]].iloc[0]
precision_check = evaluate_precision_collapse(overall_winner, overall_winner_default)

best_models_df["is_overall_winner"] = best_models_df["model"].eq(overall_winner["model"]) & best_models_df["threshold"].eq(overall_winner["threshold"])
best_models_df["precision_collapse_vs_050"] = False
best_models_df["precision_ratio_vs_050"] = np.nan
best_models_df["precision_check_message"] = "No precision collapse detected relative to threshold 0.50."

winner_mask = best_models_df["is_overall_winner"]
best_models_df.loc[winner_mask, "precision_collapse_vs_050"] = precision_check["precision_collapse_vs_050"]
best_models_df.loc[winner_mask, "precision_ratio_vs_050"] = precision_check["precision_ratio_vs_050"]
best_models_df.loc[winner_mask, "precision_check_message"] = precision_check["precision_check_message"]

results_df.to_csv(RESULTS_PATH, index=False)
auc_logloss_df.to_csv(AUC_LOGLOSS_PATH, index=False)
best_models_df.to_csv(BEST_MODELS_PATH, index=False)

run_finished_at = pd.Timestamp.now("UTC")
run_duration_seconds = (run_finished_at - run_started_at).total_seconds()

print("AUC and LogLoss per model:")
print(auc_logloss_df.to_string(index=False))

print("\nTop 10 by F1:")
print(top10_f1_df.to_string(index=False))

print("\nTop 10 by MCC:")
print(top10_mcc_df.to_string(index=False))

print("\nThreshold 0.50 summary:")
print(default_threshold_df.to_string(index=False))

print("\nRecommended winner (primary = F1, tie-breaker = MCC):")
print(pd.DataFrame([overall_winner]).to_string(index=False))

print("\nPrecision sanity check versus threshold 0.50:")
print(pd.DataFrame([precision_check]).to_string(index=False))

print("\nSaved files:")
for output_path in [RESULTS_PATH, AUC_LOGLOSS_PATH, BEST_MODELS_PATH]:
    print(f"- {output_path}")

print(f"\nRun duration (seconds): {run_duration_seconds:.3f}")
```

## 7. Validate Behavior with Assertions

These checks confirm that the notebook used the expected shape, binary target, exact threshold grid, and required output files.

```python
assert df.shape == EXPECTED_SHAPE
assert set(y_true.unique().tolist()) == {0, 1}
assert len(THRESHOLDS) == 76
assert results_df["model"].nunique() == len(MODEL_METADATA)
assert results_df.shape[0] == len(MODEL_METADATA) * len(THRESHOLDS)
assert default_threshold_df.shape[0] == len(MODEL_METADATA)
assert set(auc_logloss_df["model"].tolist()) == set(MODEL_METADATA.keys())
assert set(best_models_df["model"].tolist()) == set(MODEL_METADATA.keys())
for expected_path in [RESULTS_PATH, AUC_LOGLOSS_PATH, BEST_MODELS_PATH]:
    assert expected_path.exists(), f"Expected output file missing: {expected_path}"

print("All validation checks passed.")
```
